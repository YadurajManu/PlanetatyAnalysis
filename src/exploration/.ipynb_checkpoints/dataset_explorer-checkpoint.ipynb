# Astronomical Dataset Explorer 🌟🔭

Let's dive into our synthetic astronomical datasets!

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load datasets
exoplanets = pd.read_csv('../../data/raw/quick_exoplanets.csv')
stars = pd.read_csv('../../data/raw/quick_stars.csv')

# Display basic information
print("Exoplanet Dataset Info:")
print(exoplanets.info())

print("\nStar Dataset Info:")
print(stars.info())
```

## Exoplanet Exploration 🪐

```python
# Visualize planet radius distribution
plt.figure(figsize=(10, 6))
plt.hist(exoplanets['planet_radius'], bins=20, edgecolor='black')
plt.title('Distribution of Exoplanet Radii')
plt.xlabel('Planet Radius (Jupiter Radii)')
plt.ylabel('Number of Planets')
plt.show()

# Scatter plot of planet mass vs orbital period
plt.figure(figsize=(10, 6))
plt.scatter(exoplanets['orbital_period'], exoplanets['planet_mass'], alpha=0.6)
plt.title('Orbital Period vs Planet Mass')
plt.xlabel('Orbital Period (Days)')
plt.ylabel('Planet Mass (Jupiter Masses)')
plt.show()

# Habitable zone planets
habitable_planets = exoplanets[exoplanets['habitable_zone']]
print(f"\nPotentially Habitable Planets: {len(habitable_planets)} out of {len(exoplanets)}")
print(habitable_planets[['planet_name', 'star_name', 'planet_radius', 'star_temperature']].head())
```

## Stellar Analysis ⭐

```python
# Spectral type distribution
plt.figure(figsize=(10, 6))
stars['spectral_type'].value_counts().plot(kind='bar')
plt.title('Distribution of Stellar Spectral Types')
plt.xlabel('Spectral Type')
plt.ylabel('Number of Stars')
plt.show()

# Temperature vs Luminosity
plt.figure(figsize=(10, 6))
plt.scatter(stars['temperature'], stars['luminosity'], alpha=0.6)
plt.title('Star Temperature vs Luminosity')
plt.xlabel('Temperature (Kelvin)')
plt.ylabel('Luminosity (Solar Luminosity)')
plt.xscale('log')
plt.yscale('log')
plt.show()

# Star coordinate visualization
plt.figure(figsize=(10, 6))
plt.scatter(stars['right_ascension'], stars['declination'], alpha=0.6)
plt.title('Star Positions in Sky')
plt.xlabel('Right Ascension (degrees)')
plt.ylabel('Declination (degrees)')
plt.show()
```
